```python
# initial conditions
Glc_ext = 0.0
Glc     = 0.0

Glc_in  = 2.0
dt      = 0.01
steps   = 500

history = []

for _ in range(steps):
    dGlc_ext = kinetics.dGlc_ext_dt(Glc_ext, Glc_in)
    dGlc     = kinetics.dGlc_dt(Glc, Glc_ext)

    Glc_ext += dt * dGlc_ext
    Glc     += dt * dGlc

    history.append([Glc_ext, Glc])

history = np.array(history)

plt.plot(history[:,0], label="Glc_ext")
plt.plot(history[:,1], label="Glc")
plt.ylim(-0.01, 0.3)
plt.legend()
plt.show()

print("Final Glc_ext:", history[-1, 0])
print("Final Glc:", history[-1, 1])
```

```python
# --- initial conditions ---
Glc_ext  = 0.0
Glc      = 0.0
Glc6P    = 0.0
Fru6P    = 0.0
Fru16P2  = 0.0
Pyruvate = 0.0
AcCoA    = 0.0   # shared pool

Glc_in   = 2.0

dt    = 0.01
steps = 500

history = []

for _ in range(steps):
    Glc_ext  += dt * kinetics.dGlc_ext_dt(Glc_ext, Glc_in)
    Glc      += dt * kinetics.dGlc_dt(Glc, Glc_ext)
    Glc6P    += dt * kinetics.dGlc6P_dt(Glc6P, Glc)
    Fru6P    += dt * kinetics.dFru6P_dt(Fru6P, Glc6P, Cit=0.0)
    Fru16P2  += dt * kinetics.dFru16P2_dt(Fru16P2, Fru6P, Cit=0.0)
    Pyruvate += dt * kinetics.dPyruvate_dt(Pyruvate, Fru16P2, AcCoA)
    AcCoA    += dt * kinetics.dAcetylCoA_dt(AcCoA, Pyruvate, LCFA_CoA_mito=0.0, ATP_draw=0.0)

    history.append([Glc_ext, Glc, Glc6P, Fru6P, Fru16P2, Pyruvate, AcCoA])

history = np.array(history)

plt.plot(history[:,0], label="Glc_ext")
plt.plot(history[:,1], label="Glc")
plt.plot(history[:,2], label="Glc6P")
plt.plot(history[:,3], label="Fru6P")
plt.plot(history[:,4], label="Fru16P2")
plt.plot(history[:,5], label="Pyruvate")
plt.plot(history[:,6], label="AcCoA")
plt.ylim(-0.01,0.3)
plt.legend()
plt.show()

print("Final Glc_ext:", history[-1, 0])
print("Final Glc:",     history[-1, 1])
print("Final Glc6P:",   history[-1, 2])
print("Final Fru6P:",   history[-1, 3])
print("Final Fru16P2:", history[-1, 4])
print("Final Pyruvate:",history[-1, 5])
print("Final AcCoA:",   history[-1, 6])
```

```python
# ============================
# INITIAL CONDITIONS
# ============================

Glc_ext    = 0.0
Glc        = 0.0
Glc6P      = 0.0
Fru6P      = 0.0
Fru16P2    = 0.0
Pyruvate   = 0.0
AcCoA      = 0.0    # <-- unified pool
Cit        = 0.0


# ============================
# USER INPUTS
# ============================

Glc_in    = 2.0


# ============================
# INTEGRATION SETTINGS
# ============================

dt    = 0.01
steps = 500

history = []


# ============================
# MAIN INTEGRATION LOOP
# ============================

for _ in range(steps):

    Glc_ext   += dt * kinetics.dGlc_ext_dt(Glc_ext, Glc_in)
    Glc       += dt * kinetics.dGlc_dt(Glc, Glc_ext)
    Glc6P     += dt * kinetics.dGlc6P_dt(Glc6P, Glc)
    Fru6P     += dt * kinetics.dFru6P_dt(Fru6P, Glc6P, Cit)
    Fru16P2   += dt * kinetics.dFru16P2_dt(Fru16P2, Fru6P, Cit)
    Pyruvate  += dt * kinetics.dPyruvate_dt(Pyruvate, Fru16P2, AcCoA)
    # unified AcCoA pool: glucose + fat contributions
    AcCoA    += dt * kinetics.dAcetylCoA_dt(
        AcCoA,
        Pyruvate,
        LCFA_CoA_mito=0.0,
        ATP_draw=0.0
    )

    # citrate from unified AcCoA pool
    Cit      += dt * kinetics.dCit_dt(Cit, AcCoA, Glc)

    history.append([Glc_ext, Glc, Glc6P, Fru6P, Fru16P2, Pyruvate, AcCoA, Cit])


history = np.array(history)


# ============================
# PLOT RESULTS
# ============================

plt.plot(history[:,0], label="Glc_ext")
plt.plot(history[:,1], label="Glc")
plt.plot(history[:,2], label="Glc6P")
plt.plot(history[:,3], label="Fru6P")
plt.plot(history[:,4], label="Fru16P2")
plt.plot(history[:,5], label="Pyruvate")
plt.plot(history[:,6], label="AcCoA")
plt.plot(history[:,7], label="Cit")
plt.legend()
plt.ylim(-0.01,0.3)
plt.show()


# ============================
# FINAL VALUES
# ============================

print("Final Glc_ext:",   history[-1, 0])
print("Final Glc:",       history[-1, 1])
print("Final Glc6P:",     history[-1, 2])
print("Final Fru6P:",     history[-1, 3])
print("Final Fru16P2:",   history[-1, 4])
print("Final Pyruvate:",  history[-1, 5])
print("Final AcCoA:", history[-1, 6])
print("Final Cit:",       history[-1, 7])
```

```python
# ----------------------------
# Shared settings
# ----------------------------
Glc_in   = 2.0
dt       = 0.01
steps    = 500

def run_case(Cit0):
    Glc_ext    = 0.0
    Glc        = 0.0
    Glc6P      = 0.0
    Fru6P      = 0.0
    Fru16P2    = 0.0
    Pyruvate   = 0.0
    AcCoA      = 0.0
    Cit        = Cit0

    Fru16P2_hist = []
    Pyr_hist     = []
    Cit_hist     = []

    for _ in range(steps):
        Glc_ext   += dt * kinetics.dGlc_ext_dt(Glc_ext, Glc_in)
        Glc       += dt * kinetics.dGlc_dt(Glc, Glc_ext)
        Glc6P     += dt * kinetics.dGlc6P_dt(Glc6P, Glc)
        Fru6P     += dt * kinetics.dFru6P_dt(Fru6P, Glc6P, Cit)
        Fru16P2   += dt * kinetics.dFru16P2_dt(Fru16P2, Fru6P, Cit)
        Pyruvate  += dt * kinetics.dPyruvate_dt(Pyruvate, Fru16P2, AcCoA)
        AcCoA += dt * kinetics.dAcetylCoA_dt(
            AcCoA,
            Pyruvate,
            LCFA_CoA_mito=0.0,
            ATP_draw=0.0
        )
        Cit       += dt * kinetics.dCit_dt(Cit, AcCoA, Glc)

        Fru16P2_hist.append(Fru16P2)
        Pyr_hist.append(Pyruvate)
        Cit_hist.append(Cit)

    return np.array(Fru16P2_hist), np.array(Pyr_hist), np.array(Cit_hist)

# ----------------------------
# Run both cases
# ----------------------------
F16_A, Pyr_A, Cit_A = run_case(Cit0=0.0)
F16_B, Pyr_B, Cit_B = run_case(Cit0=5.0)

t = np.arange(steps) * dt

# ----------------------------
# Plot Fru16P2
# ----------------------------
plt.figure(figsize=(8,4))
plt.plot(t, F16_A, label="Fru16P2 (Cit0=0.0)")
plt.plot(t, F16_B, label="Fru16P2 (Cit0=5.0)")
plt.xlabel("time")
plt.ylabel("Fru16P2")
plt.legend()
plt.title("PFK output under low vs high citrate")
plt.show()

# ----------------------------
# Plot Pyruvate
# ----------------------------
plt.figure(figsize=(8,4))
plt.plot(t, Pyr_A, label="Pyruvate (Cit0=0.0)")
plt.plot(t, Pyr_B, label="Pyruvate (Cit0=5.0)")
plt.xlabel("time")
plt.ylabel("Pyruvate")
plt.legend()
plt.title("Pyruvate under low vs high citrate")
plt.show()

# Optional: final values
print("Final Fru16P2 (Cit0=0.0):", F16_A[-1])
print("Final Fru16P2 (Cit0=5.0):", F16_B[-1])
print("Final Pyruvate (Cit0=0.0):", Pyr_A[-1])
print("Final Pyruvate (Cit0=5.0):", Pyr_B[-1])
print("Final Cit (Cit0=0.0):", Cit_A[-1])
print("Final Cit (Cit0=5.0):", Cit_B[-1])
```

```python
# --- LCFA-CoA_cyto verification test ---

# initial conditions
LCFA_ext      = 0.0     # mM, extracellular LCFA
LCFA_in       = 0.2     # mM/min, supply rate
LCFA_CoA_cyto = 0.0     # mM
Mal           = 0.0     # mM
Glc           = 0.0     # mM

dt    = 0.01
steps = 500

history = []

for _ in range(steps):
    # LCFA_ext (CD36 uptake)
    LCFA_ext += dt * kinetics.dLCFA_ext_dt(LCFA_ext, LCFA_in)

    # LCFA-CoA_cyto (CD36 + FAS - CPT1 - TAG)
    LCFA_CoA_cyto += dt * kinetics.dLCFA_CoA_cyto_dt(
        LCFA_ext,
        LCFA_CoA_cyto,
        Mal,
        Glc
    )

    # append all states
    history.append([
        LCFA_ext,
        LCFA_CoA_cyto,
        Mal,
        Glc
    ])

history = np.array(history)

# --- plot ---
plt.plot(history[:,0], label="LCFA_ext")
plt.plot(history[:,1], label="LCFA-CoA_cyto")
plt.legend()
plt.show()

# --- final values ---
print("Final LCFA_ext:      ", history[-1,0])
print("Final LCFA-CoA_cyto: ", history[-1,1])
```

```python
# --- LCFA_ext + LCFA-CoA_cyto + LCFA-CoA_mito + AcCoA_Fat verification ---

# initial conditions
LCFA_ext       = 0.0
LCFA_in        = 0.2
LCFA_CoA_cyto  = 0.0
LCFA_CoA_mito  = 0.0
AcCoA          = 0.0
Mal            = 0.0
Glc            = 0.0
ATP_draw       = 0.0

dt    = 0.01
steps = 500

history = []

for _ in range(steps):
    # LCFA_ext (CD36 uptake)
    LCFA_ext += dt * kinetics.dLCFA_ext_dt(LCFA_ext, LCFA_in)

    # LCFA-CoA_cyto (CD36 + FAS - CPT1 - TAG)
    LCFA_CoA_cyto += dt * kinetics.dLCFA_CoA_cyto_dt(
        LCFA_ext,
        LCFA_CoA_cyto,
        Mal,
        Glc
    )

    # LCFA-CoA_mito (CPT1 - beta-ox)
    LCFA_CoA_mito += dt * kinetics.dLCFA_CoA_mito_dt(
        LCFA_CoA_cyto,
        LCFA_CoA_mito,
        Mal
    )

    # unified AcCoA pool: only fat contributes here
    AcCoA += dt * kinetics.dAcetylCoA_dt(
        AcCoA,
        Pyruvate=0.0,
        LCFA_CoA_mito=LCFA_CoA_mito,
        ATP_draw=ATP_draw
    )

    # store all states
    history.append([
        LCFA_ext,
        LCFA_CoA_cyto,
        LCFA_CoA_mito,
        AcCoA
    ])

history = np.array(history)

# --- plot ---
plt.plot(history[:,0], label="LCFA_ext")
plt.plot(history[:,1], label="LCFA-CoA_cyto")
plt.plot(history[:,2], label="LCFA-CoA_mito")
plt.plot(history[:,3], label="AcCoA")
plt.legend()
plt.show()

# --- final values ---
print("Final LCFA_ext:       ", history[-1,0])
print("Final LCFA-CoA_cyto:  ", history[-1,1])
print("Final LCFA-CoA_mito:  ", history[-1,2])
print("Final AcCoA:          ", history[-1,3])
```

### Full integration!

```python
# ============================
# INITIAL CONDITIONS
# ============================

# --- glucose side ---
Glc_ext    = 0.0
Glc        = 0.0
Glc6P      = 0.0
Fru6P      = 0.0
Fru16P2    = 0.0
Pyruvate   = 0.0

# --- fat side ---
LCFA_ext       = 0.0
LCFA_CoA_cyto  = 0.0
LCFA_CoA_mito  = 0.0

# --- shared mitochondrial AcCoA ---
AcCoA = 0.0

# --- regulators ---
Cit = 0.0
Mal = 0.0

# ============================
# USER INPUTS
# ============================

Glc_in    = 2.0
LCFA_in   = 0.2
ATP_draw  = 0.0

# ============================
# INTEGRATION SETTINGS
# ============================

dt    = 0.01
steps = 1000

history = []

# ============================
# MAIN INTEGRATION LOOP
# ============================

for _ in range(steps):

    # --- glucose chain ---
    Glc_ext   += dt * kinetics.dGlc_ext_dt(Glc_ext, Glc_in)
    Glc       += dt * kinetics.dGlc_dt(Glc, Glc_ext)
    Glc6P     += dt * kinetics.dGlc6P_dt(Glc6P, Glc)
    Fru6P     += dt * kinetics.dFru6P_dt(Fru6P, Glc6P, Cit)
    Fru16P2   += dt * kinetics.dFru16P2_dt(Fru16P2, Fru6P, Cit)
    Pyruvate  += dt * kinetics.dPyruvate_dt(Pyruvate, Fru16P2, AcCoA)

    # --- fat chain ---
    LCFA_ext      += dt * kinetics.dLCFA_ext_dt(LCFA_ext, LCFA_in)
    LCFA_CoA_cyto += dt * kinetics.dLCFA_CoA_cyto_dt(
        LCFA_ext,
        LCFA_CoA_cyto,
        Mal,
        Glc
    )
    LCFA_CoA_mito += dt * kinetics.dLCFA_CoA_mito_dt(
        LCFA_CoA_cyto,
        LCFA_CoA_mito,
        Mal
    )

    # ---- common AcCoA pool ----
    AcCoA     += dt * kinetics.dAcetylCoA_dt(AcCoA, Pyruvate, LCFA_CoA_mito, ATP_draw)

    # --- citrate (from both AcCoA pools) ---
    Cit += dt * kinetics.dCit_dt(
        Cit,
        AcCoA,
        Glc
    )

    # --- malonyl (ACC activation by citrate + glucose) ---
    Mal += dt * kinetics.dMal_dt(
        Cit,
        Mal,
        Glc
    )

    history.append([
        Glc_ext, Glc, Glc6P, Fru6P, Fru16P2,
        Pyruvate,
        LCFA_ext, LCFA_CoA_cyto, LCFA_CoA_mito, AcCoA,
        Cit, Mal
    ])

history = np.array(history)

##### PLOTTING #####
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# ============================
# LEFT PANEL — Glucose Pathway
# ============================

axes[0].plot(history[:,0], label="Glc_ext")
axes[0].plot(history[:,1], label="Glc")
axes[0].plot(history[:,2], label="Glc6P")
axes[0].plot(history[:,3], label="Fru6P")
axes[0].plot(history[:,4], label="Fru16P2")
axes[0].plot(history[:,5], label="Pyruvate")
axes[0].plot(history[:,9], label="AcCoA")   # unified pool
axes[0].plot(history[:,10], label="Cit")

axes[0].set_title("Glucose Pathway")
axes[0].set_xlabel("Time")
axes[0].set_ylabel("Concentration (mM)")
axes[0].set_ylim(-0.01, 0.5)
axes[0].legend(loc="upper left")


# ============================
# RIGHT PANEL — Fat Pathway
# ============================

axes[1].plot(history[:,6], label="LCFA_ext")
axes[1].plot(history[:,7], label="LCFA-CoA_cyto")
axes[1].plot(history[:,8], label="LCFA-CoA_mito")
axes[1].plot(history[:,9], label="AcCoA")   # unified pool
axes[1].plot(history[:,10], label="Cit")
axes[1].plot(history[:,11], label="Mal")

axes[1].set_title("Fat Pathway")
axes[1].set_xlabel("Time")
axes[1].set_ylabel("Concentration (mM)")
axes[1].set_ylim(-0.01, 0.1)
axes[1].legend(loc="upper left")

plt.tight_layout()
plt.show()


# ============================
# FINAL VALUES
# ============================

print("Final Glc_ext:",      history[-1,0])
print("Final Glc:",          history[-1,1])
print("Final Glc6P:",        history[-1,2])
print("Final Fru6P:",        history[-1,3])
print("Final Fru16P2:",      history[-1,4])
print("Final Pyruvate:",     history[-1,5])

print("Final LCFA_ext:",      history[-1,6])
print("Final LCFA-CoA_cyto:", history[-1,7])
print("Final LCFA-CoA_mito:", history[-1,8])
print("Final AcCoA:",         history[-1,9])

print("Final Cit:",           history[-1,10])
print("Final Mal:",           history[-1,11])
```
