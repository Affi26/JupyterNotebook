# Comprehensive NHST Example Problem

A researcher wants to investigate whether a new study technique improves exam scores.

Historically, students at a university score an average of

$$
\mu_0 = 70
$$

points on a standardized exam, with a known population standard deviation

$$
\sigma = 15.
$$

The researcher recruits

$$
n = 36
$$

students and teaches them the new technique.

After the intervention, the sample mean exam score is

$$
\bar{x} = 75.
$$

Assume scores are approximately normally distributed.

The researcher tests

$$
H_0:\mu = 70
$$

against

$$
H_A:\mu > 70
$$

using significance level

$$
\alpha = 0.05.
$$

# Answers

### A1

According to CLT, sample mean will be normally distributed under the null hypothesis as $\mu \sim \mathcal{N}(\mu_0, SE_\mu)$
with
$$
\mu_0 = 70 
$$
and
$$
SE_\mu = \frac{\sigma}{\sqrt{n}} = \frac{15}{\sqrt{36}} \approx 2.5
$$

### A2

$$
SE_\mu = \frac{\sigma}{\sqrt{n}} = \frac{15}{\sqrt{36}} \approx 2.5
$$

### A3

$$
P(\bar{x} > 75 | H_0) = \int_{75}^\infty \mathcal{N}(70, 2.5) dx \approx 0.02
$$

The interpretation is exactly what the question is.
The probability of observing a sample mean value of 75 or greater if we assume the true underlying sample mean distribution is defined by $\mathcal{N}(70, 2.5)$

### B1

This is a bit repetitive, but the test statistic would be a z-score:
$$
z = \frac{\bar{x} - \mu}{\sigma/\sqrt{n}} = \frac{5}{2.5} = 2
$$
(since $n = 36$ is sufficient for a good normal approximation, and the population standard deviation is known)

### B2

Since $Z \sim \mathcal{N}(0,1)$ we get
$$
p = \int_z^\infty \mathcal{N}(0,1) dx = \int_2^\infty \mathcal{N}(0,1) dx \approx 0.02
$$

### B3

Based on the p-value and the $\alpha$ level, since $p < \alpha$, the result is statistically significant.
We can reject the null hypothesis in favor of the alternative hypothesis.

### B4

Again a bit repetitive... it means the probability of observing a z-score of 2 or higher if we assume the true underlying distribution is
$Z \sim \mathcal{N}(0,1)$.

### C1

$$
95\%CI = \bar{x} \pm 1.96 \cdot SE = [70.1, 79.9]
$$

### C2

Given that the lower bound of the confidence interval is above the null hypothesis criterion $\mu = 70$,
the CI seems to agree with the conclusion of a statistically significant result for the sample mean.

### D1

$$
d = \frac{75-70}{15} \approx 0.33
$$

### D2

According to Cohen's traditional thresholds (0.2, 0.5, 0.8),
$d = 0.33$ is a small effect size.

### D3

Statistical significance concerns inferences and uncertainty about parameter estimations,
while effect sizes concerns the actual observations and sample/population distributions.

### E1

Type I error would mean the probability of falsely rejecting H0 when H0 is true.

### E2

Type II error would mean the probability of failing to reject H0 when H1 is true.

### E3

Type I error.

### F1

Using the same $SE = 2.5$ as before, we would need a sample mean of $\bar{x} > 70 + 1.64 \cdot 2.5 \approx 74.1$ (one-sided) to reject $H_0$.

### F2

Given the true distribution of the mean is $\mathcal{N}(75, 2.5)$ we would calculate the probability as
$$
\int_{74.1}^\infty \mathcal{N}(75, 2.5) dx \approx 0.64
$$
This is equal to the power:
$$
\text{Power} = 1 - \beta = 64\%
$$

### F3

Probability of Type II error is
$$
\beta = 1 - 0.64 = 0.36 = 36\%
$$

### F4

Kind of repetitive...

### G1

SE would decrease, test statistic would increase, p-value would decrease, power would increase.

### G2

- Yes
- Yes

### G3

P-values only imply the parameter value is not produced "due to dumb chance".
But p-values do not inform on real-world utility, which is informed much better by effect sizes.
Effect sizes also have more information about the reproducibility of results than p-values alone.
